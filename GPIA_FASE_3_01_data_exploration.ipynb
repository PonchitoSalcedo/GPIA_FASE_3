```python
# -*- coding: utf-8 -*-
"""01_data_exploration.ipynb

Automatically generated by Colaboratory.
"""

# 📊 Exploración de Datos - California Housing
# Pipeline Automatizado de IA con MLOps y GitOps
# Fase 1: Análisis Exploratorio de Datos (EDA)

# Instalación de dependencias
get_ipython().system('pip install pandas numpy matplotlib seaborn scikit-learn --quiet')
get_ipython().system('pip install pandas-profiling --quiet')

print("✅ Dependencias instaladas")

# Importación de librerías
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

print("✅ Librerías importadas correctamente")
print(f"Versión de pandas: {pd.__version__}")

# Carga de datos
housing = fetch_california_housing()
df = pd.DataFrame(housing.data, columns=housing.feature_names)
df['MedHouseVal'] = housing.target

print("📊 Información del dataset:")
print(f"Número de registros: {df.shape[0]}")
print(f"Número de features: {df.shape[1]}")
print(f"\nColumnas: {df.columns.tolist()}")
print(f"\nDescripción de las features:")
print(housing.DESCR[:500] + "...")

# Análisis estadístico
print("📈 Estadísticas descriptivas:")
display(df.describe())

# Verificación de valores nulos
print("\n🔍 Valores nulos por columna:")
display(df.isnull().sum())

# Visualización de distribución de variables
fig, axes = plt.subplots(3, 3, figsize=(15, 12))
axes = axes.ravel()

for idx, col in enumerate(df.columns):
    axes[idx].hist(df[col], bins=50, edgecolor='black', alpha=0.7)
    axes[idx].set_title(f'Distribución de {col}')
    axes[idx].set_xlabel(col)
    axes[idx].set_ylabel('Frecuencia')

plt.tight_layout()
plt.savefig('distribucion_variables.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Gráfico guardado como 'distribucion_variables.png'")

# Matriz de correlación
plt.figure(figsize=(12, 10))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm',
            square=True, linewidths=0.5)
plt.title('Matriz de Correlación del Dataset California Housing')
plt.tight_layout()
plt.savefig('matriz_correlacion.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Gráfico guardado como 'matriz_correlacion.png'")

# Análisis de outliers mediante boxplots
fig, axes = plt.subplots(3, 3, figsize=(15, 12))
axes = axes.ravel()

for idx, col in enumerate(df.columns):
    axes[idx].boxplot(df[col])
    axes[idx].set_title(f'Boxplot de {col}')
    axes[idx].set_ylabel(col)

plt.tight_layout()
plt.savefig('boxplots_outliers.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Gráfico guardado como 'boxplots_outliers.png'")

# Análisis de relaciones con la variable objetivo
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
axes = axes.ravel()

for idx, col in enumerate(df.columns[:-1]):
    axes[idx].scatter(df[col], df['MedHouseVal'], alpha=0.5, s=1)
    axes[idx].set_xlabel(col)
    axes[idx].set_ylabel('MedHouseVal')
    axes[idx].set_title(f'{col} vs MedHouseVal')

plt.tight_layout()
plt.savefig('relaciones_target.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Gráfico guardado como 'relaciones_target.png'")

# 🔧 PARTE CORREGIDA - Reporte de calidad de datos (sin pandas-profiling)
print("📊 Generando reporte de calidad de datos...")

# Generar reporte HTML simple con estadísticas
with open('data_profile_report.html', 'w') as f:
    f.write("""
    <html>
    <head>
        <title>Reporte de Datos - California Housing</title>
        <style>
            body { font-family: Arial, sans-serif; margin: 20px; }
            table { border-collapse: collapse; width: 100%; }
            th, td { border: 1px solid #ddd; padding: 8px; text-align: left; }
            th { background-color: #4CAF50; color: white; }
            .titulo { color: #2c3e50; }
            .resumen { background-color: #f0f8ff; padding: 10px; border-radius: 5px; }
        </style>
    </head>
    <body>
        <h1 class="titulo">📊 Reporte de Datos - California Housing</h1>
        <div class="resumen">
            <h2>Resumen del Dataset</h2>
            <p><strong>Número de registros:</strong> """ + str(df.shape[0]) + """</p>
            <p><strong>Número de features:</strong> """ + str(df.shape[1]) + """</p>
            <p><strong>Columnas:</strong> """ + ', '.join(df.columns.tolist()) + """</p>
        </div>
        <h2>📈 Estadísticas Descriptivas</h2>
        """ + df.describe().to_html() + """
        <h2>🔍 Información de Tipos de Datos</h2>
        """ + df.dtypes.to_frame('Tipo de Dato').to_html() + """
        <h2>📊 Valores Nulos</h2>
        """ + df.isnull().sum().to_frame('Valores Nulos').to_html() + """
    </body>
    </html>
    """)

print("✅ Reporte de calidad generado como 'data_profile_report.html'")

# Guardar datos procesados para notebooks posteriores
import joblib

# Guardar datos completos
joblib.dump(df, 'california_housing_data.pkl')
print("✅ Datos guardados como 'california_housing_data.pkl'")

# Guardar separados
X = df.drop('MedHouseVal', axis=1)
y = df['MedHouseVal']
joblib.dump(X, 'X_data.pkl')
joblib.dump(y, 'y_data.pkl')
print("✅ Features y target guardados como 'X_data.pkl' y 'y_data.pkl'")

# Guardar nombres de columnas
joblib.dump(df.columns.tolist(), 'column_names.pkl')
print("✅ Nombres de columnas guardados como 'column_names.pkl'")

# Resumen final
print("\n" + "="*60)
print("📊 RESUMEN DEL ANÁLISIS EXPLORATORIO")
print("="*60)
print(f"\n📌 Dataset: California Housing")
print(f"📊 Registros: {df.shape[0]}")
print(f"📋 Features: {df.shape[1]}")
print(f"🎯 Target: MedHouseVal (Valor medio de vivienda)")
print(f"\n📈 Rango de precios: ${df['MedHouseVal'].min()*100000:.0f} - ${df['MedHouseVal'].max()*100000:.0f}")
print(f"💰 Precio promedio: ${df['MedHouseVal'].mean()*100000:.0f}")

print("\n📁 ARCHIVOS GENERADOS:")
print("-"*40)
archivos = [
    'distribucion_variables.png',
    'matriz_correlacion.png',
    'boxplots_outliers.png',
    'relaciones_target.png',
    'data_profile_report.html',
    'california_housing_data.pkl',
    'X_data.pkl',
    'y_data.pkl',
    'column_names.pkl'
]
for archivo in archivos:
    import os
    if os.path.exists(archivo):
        size = os.path.getsize(archivo) / 1024
        print(f"  ✅ {archivo} ({size:.1f} KB)")
    else:
        print(f"  ❌ {archivo} (NO ENCONTRADO)")

print("\n🎯 PRÓXIMO PASO: Ejecutar 02_model_training.ipynb")

# Código para descargar todos los archivos
print("\n📥 Descargando archivos generados...")
from google.colab import files
import os
import zipfile

# Crear ZIP con todos los archivos
with zipfile.ZipFile('notebook1_artefactos.zip', 'w') as zipf:
    archivos = [
        'distribucion_variables.png',
        'matriz_correlacion.png',
        'boxplots_outliers.png',
        'relaciones_target.png',
        'data_profile_report.html',
        'california_housing_data.pkl',
        'X_data.pkl',
        'y_data.pkl',
        'column_names.pkl'
    ]
    for archivo in archivos:
        if os.path.exists(archivo):
            zipf.write(archivo)
            print(f"  📦 {archivo} añadido al ZIP")

# Descargar ZIP
files.download('notebook1_artefactos.zip')
print("✅ Descarga completada!")
```
